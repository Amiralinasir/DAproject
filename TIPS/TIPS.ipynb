```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

```python
plt.hist(df["tip"], bins=70)
plt.xlabel("tip")
plt.ylabel("count")
```

```python
plt.hist(df[df["sex"]=="Male"]["tip"],bins=70)
plt.xlabel("tip")
plt.ylabel("count")
```
