```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("taxis")
df
```

```python
plt.hist(df["total"], bins=70)
plt.xlabel("total")
plt.ylabel("count")
```

```python
plt.hist(
    df[(df["distance"] >=0) & (df["distance"] <= 1)]["total"],
    bins=70
)

plt.xlabel("Total")
plt.ylabel("Frequency")
```

```python
plt.hist(
    df[(df["distance"] >1) & (df["distance"] <= 2)]["total"],
    bins=70
)

plt.xlabel("Total")
plt.ylabel("Frequency")
```

```python
plt.hist(
    df[(df["distance"] >2) & (df["distance"] <= 3)]["total"],
    bins=70
)

plt.xlabel("Total")
plt.ylabel("Frequency")
```

```python
plt.hist(
    df[(df["distance"] >3)]["total"],
    bins=70
)

plt.xlabel("Total")
plt.ylabel("Frequency")
```

```python
plt.hist(
    df[df["payment"] == "credit card"]["total"],
    bins=70
)

plt.xlabel("Total")
plt.ylabel("Count")
```

```python
plt.hist(
    df[df["payment"] == "cash"]["total"],
    bins=70
)

plt.xlabel("Total")
plt.ylabel("Count")
```
